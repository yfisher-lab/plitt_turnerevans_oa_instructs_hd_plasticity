```python
import numpy as np
from matplotlib import pyplot as plt
```

```python
def von_mises_pdf(x, mu, kappa):
    """Calculate the von Mises probability density function."""
    return np.exp(kappa * np.cos(x - mu)) / (2 * np.pi * np.i0(kappa))
mus = np.arange(0, 2 * np.pi, np.pi / 4)  # Mean directions
kappa = 4.0  # Concentration parameter
```

```python
for i, mu in enumerate(mus):
    fig = plt.figure()
    ax = fig.add_subplot(111, polar=True)

    
    x = np.linspace(-np.pi, np.pi, 100)
    y = von_mises_pdf(x, mu, kappa)
    ax.plot(x, y, color=plt.cm.hsv(mu / (2 * np.pi)))  # Color based on mean direction
    ax.set_yticks([])  # Remove radial ticks
    ax.set_xticks(np.array([0, np.pi/4, np.pi/2, 3*np.pi/4, np.pi, 5*np.pi/4, 3*np.pi/2, 7*np.pi/4, ]))  # Set angular ticks
    ax.set_xticklabels([0, 'π/4', 'π/2', '3π/4', 'π', '-3π/4','-π/2', '-π/4',  ])  # Set angular tick labels
    ax.set_ylabel('Firing Rate')
```
